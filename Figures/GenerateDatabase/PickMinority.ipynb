```python
import pandas as pd
from pathlib import Path
import sys
sys.path.append("../../tools")
import functions as func
import prepare as prep

import importlib
importlib.reload(prep)
importlib.reload(func)

def batch_clean_csv(input_dir, output_dir):
    """
    Batch process all CSV files in a folder
    """
    input_dir = Path(input_dir)
    output_dir = Path(output_dir)

    # Create output directory if it does not exist
    output_dir.mkdir(parents=True, exist_ok=True)

    # Iterate over all CSV files
    for file_path in input_dir.glob("*.csv"):
        n_real, NumSpecies = func.get_species_info(file_path)
        df1 = pd.read_csv(file_path)
        # First extract the Xnum portion of the minority class
        df2 = func.pick_minority(NumSpecies, 3 + 2 * NumSpecies, df=df1)
        print(f"\n✅ Processed {file_path.name}: {len(df2)} minority samples selected")
        # Then restore the padded Xnum to original unpadded form
        Xnum_pad = df2.values
        Xnum = prep.recovery_Xnum(Xnum_pad, n_real, NumSpecies)
        # Resample
        Xnum_intensive = func.intensive_sampling_Xnum(Xnum, n_real, num_samples=10, sigma_mu=0.03, sigma_gamma=0.3, sigma_D=3e-12)
        df3 = pd.DataFrame(Xnum_intensive)
        # Chunk the dataset for parallel processing
        n_chunks = 256
        func.split_csv(n_real, NumSpecies, df3, output_dir, n_chunks)
        print(f"✅ Split {file_path.name} into {n_chunks} chunks and saved to {output_dir}")

    print("\n✅ All files processed!")

batch_clean_csv(
    input_dir="../../Databases/Ver1.9/ver1.9.6_upto20_sp3to9/",
    output_dir="../../Databases/Ver1.9/ver1.9.6_upto20_sp3to9_intensive/"
    # input_dir="../../Databases/Ver1.9/test_raw/",
    # output_dir="../../Databases/Ver1.9/test_split/"
)
```
