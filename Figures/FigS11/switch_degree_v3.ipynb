```python
load_dir5 = Path(f"../../saved_models/swi_species_15fold/sp3to20_SS_transpt3_r13")
smodel = func.load_saved_model_SS(load_dir5, device)
spredictor = func.Predictor(smodel)
MaxNum = smodel["NumSpecies"]
featgam2 = MaxNum**2-MaxNum
```

```python
df = pd.read_csv("../Fig4/gamma5.csv")
```

```python
import numpy as np

# Filter samples by density
#sub_df = sub_df.sample(n=100, random_state=42)
#sub_df = df[df["density"]<0.33].copy()
#sub_df = df[(df["density"]>=0.33) & (df["density"]<0.66)].copy()
sub_df = df[df["density"] >= 0.66].copy()

datasize = len(sub_df)

# Extract input features
Xnum_pad = sub_df.iloc[:, :(MaxNum**2 + MaxNum)].values

# Preprocess input features
Xg, XA, Xd = prep.process_narray_X(Xnum_pad, MaxNum, featgam2, smodel["scaler_g"], smodel["scaler_d"])

# Predict switching species
swi_species = spredictor.predict_array_SS(Xnum_pad, device)

# Get true richness of each sample
richness_array = sub_df["richness"].values.astype(int)

# Original degree
switch_in_mean = []
switch_out_mean = []
nonswitch_in_mean = []
nonswitch_out_mean = []
in_mean = []
out_mean = []

# Normalized degree
switch_in_norm = []
switch_out_norm = []
nonswitch_in_norm = []
nonswitch_out_norm = []
in_norm = []
out_norm = []


for i in range(datasize):

    # True species count of current sample
    n = richness_array[i]

    # Extract the real interaction matrix and predictions
    A = XA[i, :n, :n]
    y = np.asarray(swi_species[i]).reshape(-1)[:n]

    # Remove self-interaction
    A_no_self = A.copy()
    np.fill_diagonal(A_no_self, 0)

    threshold = 1e-8
    mask = np.abs(A_no_self) > threshold

    # Calculate in-degree and out-degree
    indegree = np.sum(mask, axis=1)
    outdegree = np.sum(mask, axis=0)

    # Normalize degree by the maximum possible degree
    if n > 1:
        indegree_norm = indegree / (n - 1)
        outdegree_norm = outdegree / (n - 1)
    else:
        indegree_norm = np.zeros(n)
        outdegree_norm = np.zeros(n)

    # Identify switching and non-switching species
    switch_mask = (y == 1)
    nonswitch_mask = (y == 0)

    # Switching species
    if np.sum(switch_mask) > 0:
        switch_in_mean.append(indegree[switch_mask])
        switch_out_mean.append(outdegree[switch_mask])

        switch_in_norm.append(indegree_norm[switch_mask])
        switch_out_norm.append(outdegree_norm[switch_mask])

    # Non-switching species
    if np.sum(nonswitch_mask) > 0:
        nonswitch_in_mean.append(indegree[nonswitch_mask])
        nonswitch_out_mean.append(outdegree[nonswitch_mask])

        nonswitch_in_norm.append(indegree_norm[nonswitch_mask])
        nonswitch_out_norm.append(outdegree_norm[nonswitch_mask])

    # All species
    in_mean.append(indegree)
    out_mean.append(outdegree)

    in_norm.append(indegree_norm)
    out_norm.append(outdegree_norm)


# Merge results across all samples
def merge_arrays(data_list):
    return np.concatenate(data_list) if data_list else np.array([])


# Original degree
switch_in_mean = merge_arrays(switch_in_mean)
switch_out_mean = merge_arrays(switch_out_mean)
nonswitch_in_mean = merge_arrays(nonswitch_in_mean)
nonswitch_out_mean = merge_arrays(nonswitch_out_mean)
in_mean = merge_arrays(in_mean)
out_mean = merge_arrays(out_mean)

# Normalized degree
switch_in_norm = merge_arrays(switch_in_norm)
switch_out_norm = merge_arrays(switch_out_norm)
nonswitch_in_norm = merge_arrays(nonswitch_in_norm)
nonswitch_out_norm = merge_arrays(nonswitch_out_norm)
in_norm = merge_arrays(in_norm)
out_norm = merge_arrays(out_norm)
```
